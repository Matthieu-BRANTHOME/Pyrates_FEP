# Notebook 06: LLM annotation

Annotation of the rest of the corpus by frontier LLMs, using the same coding form as the
experts (`campaign/2_full/full_form.json`). The prompt reproduces what an annotator sees on
the FEP platform, with few-shot examples drawn from the gold standard coded by the experts;
the remaining gold annotations serve to evaluate each model's agreement with the experts.

### Inputs:
- `data/input/cleaned/cleaned_interaction_traces.csv` (notebook 01)
- `campaign/2_full/full_form.json`
- `campaign/2_full/full_result.csv` (gold standard)
